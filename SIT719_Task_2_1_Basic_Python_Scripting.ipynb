```python
import pandas as pd
import numpy as np
```

```python
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
iris = pd.read_csv(url, header=None)

iris.head()
```

```python
iris.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'class']

print("Column names:")
print(iris.columns)

iris.head()
```

```python
print("Are there any missing values (True/False)?")
print(iris.isnull().values.any())

print("\nMissing values per column:")
print(iris.isnull().sum())
```

```python
iris.loc[10:29, 'petal_length'] = np.nan

iris.loc[10:29, ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'class']]
```

```python
print("Missing values per column after setting NaN:")
print(iris.isnull().sum())

print("\nNumber of missing values in 'petal_length' only:")
print(iris['petal_length'].isnull().sum())
```

```python
iris['petal_length'].fillna(10.0, inplace=True)

print("Missing values per column after filling NaN with 10.0:")
print(iris.isnull().sum())

iris.loc[10:29, ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'class']]
```
